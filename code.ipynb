```python
import numpy as np
import cv2 as cv
```

Background subtraction is a technique in computer vision for detecting and isolating moving objects within video sequences. It plays an important role in applications like video surveillance, traffic monitoring, gesture recognition and automatic scene analysis, where distinguishing dynamic foreground elements from a static or slowly changing background is required

```python
# Background Subtraction Using MOG2
# MOG2 includes shadow detection(shadows might appear in gray), making it more robust against lighting changes and moving objects. Moving regions appear as lighter areas in the mask
cap = cv.VideoCapture('sample_vid.mp4')
fgbg_mog = cv.createBackgroundSubtractorMOG2(history=500, detectShadows=True)

if not cap.isOpened():
    print("Error: Could not open the video file. Please check your path!")
    exit()

frame_count = 0
while True:
    ret, frame = cap.read()
    if not ret:
        print("Video is not working.")
        break
    
    scale_percent = 0.5
    width = int(frame.shape[1] * scale_percent)
    height = int(frame.shape[0] * scale_percent)
    dimension = (width, height)


    frame = cv.resize(frame, dimension, interpolation=cv.INTER_AREA)

    fgmask = fgbg_mog.apply(frame)

    cv.imshow('Foreground Mask', fgmask)
    cv.imshow('Original Video', frame)

    frame_count +=1
    if frame_count >= 50:
        pass

    if cv.waitKey(30) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```

```python
# Background Subtraction Using MOG
# We use the MOG algorithm to model each pixel as a mixture of Gaussians, distinguishing moving objects (foreground) from the background
# The resulting binary mask shows detected motion in white. We only show the first 50 frames for brevity
cap = cv.VideoCapture('sample_vid.mp4')
fgbg_mog = cv.bgsegm.createBackgroundSubtractorMOG()

if not cap.isOpened():
    print("Error: Could not open the video file. Please check your path!")
    exit()

frame_count = 0
while True:
    ret, frame = cap.read()
    if not ret:
        print("Video is not working.")
        break
    
    scale_percent = 0.5
    width = int(frame.shape[1] * scale_percent)
    height = int(frame.shape[0] * scale_percent)
    dimension = (width, height)


    frame = cv.resize(frame, dimension, interpolation=cv.INTER_AREA)

    fgmask = fgbg_mog.apply(frame)

    cv.imshow('Foreground Mask', fgmask)
    cv.imshow('Original Video', frame)

    frame_count +=1
    if frame_count >= 50:
        pass

    if cv.waitKey(30) & 0xFF == ord('q'):
        break

cap.release()
cv.destroyAllWindows()
```

```python
# Background Substraction Using GMG
# GMG requires an initial training phase before outputting valid masks. Once trained, moving objects are highlighted.
# A morphological open operation reduces noise and improves mask clarity
cap = cv.VideoCapture('sample_vid.mp4')
fgbg_gmg = cv.bgsegm.createBackgroundSubtractorGMG(initializationFrames=120, decisionThreshold=0.8)

kernel = cv.getStructuringElement(cv.MORPH_ELLIPSE, (3,3))
frame_count = 0
display_after = 130

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    scale_percent = 0.5 
    width = int(frame.shape[1] * scale_percent)
    height = int(frame.shape[0] * scale_percent)
    frame = cv.resize(frame, (width, height), interpolation=cv.INTER_AREA)

    fgmask = fgbg_gmg.apply(frame)

    frame_count += 1

    if frame_count > display_after:
        fgmask = cv.morphologyEx(fgmask, cv.MORPH_OPEN, kernel)

        cv.imshow('Original Vid', frame)
        cv.imshow('Foreground Mask', fgmask)
 
        if cv.waitKey(30) & 0xFF == ord('q'):
            break
    else:
        if cv.waitKey(1) & 0xFF == ord('q'):
            break
cap.release()
cv.destroyAllWindows()
```
